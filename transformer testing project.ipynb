# FEATURE ENGINEERING

```python
# Create time-based features
df['date']=pd.to_datetime(df['date'])
df.head()
```

```python
# Create lag feature
# Historical temperature features

df["OT_lag_1"] = df["OT"].shift(1)
df["OT_lag_6"] = df["OT"].shift(6)
df["OT_lag_12"] = df["OT"].shift(12)
df["OT_lag_24"] = df["OT"].shift(24)
df["OT_lag_48"] = df["OT"].shift(48)

df.head(55)
```

```python
# Create load lag features
load_features = ["HUFL", "HULL", "MUFL", "MULL", "LUFL", "LULL"]

for feature in load_features:
    df[f"{feature}_lag_24"] = df[feature].shift(24)
```

```python
# Create the future target
df['OT_future_24']=df['OT'].shift(-24)
df.head()
```

```python
# Remove rows with missing lag/target values
df = df.dropna().reset_index(drop=True)

print("New dataset shape:", df.shape)
```

```python
df.head()
```

```python
df.tail()
```

```python
print(df.columns.tolist())
```

```python
# Define target
target = "OT_future_24"

# Remove date and target from input features
X = df.drop(columns=["date", target])
y = df[target]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

# start train /test /split 

```python
split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=200,
    max_depth=15,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

```

```python
# make prediction 
y_pred = model.predict(X_test)
```

```python
# evaluate the model 
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

rmse = np.sqrt(mean_squared_error(y_test, y_pred))

r2 = r2_score(y_test, y_pred)


print("RMSE:", rmse)
print("R2 Score:", r2)
```

# Thermal Feature Engineering

```python
# Feature engineering
# Temperature trend features

df["OT_rolling_mean_6"] = df["OT"].rolling(6).mean()
df["OT_rolling_mean_24"] = df["OT"].rolling(24).mean()

df["OT_rolling_std_24"] = df["OT"].rolling(24).std()

df["OT_change_1h"] = df["OT"] - df["OT_lag_1"]
df["OT_change_6h"] = df["OT"] - df["OT_lag_6"]


# Load trend features

load_features = ["HUFL", "HULL", "MUFL", "MULL", "LUFL", "LULL"]

for feature in load_features:
    df[f"{feature}_rolling_mean_6"] = df[feature].rolling(6).mean()
    df[f"{feature}_rolling_mean_24"] = df[feature].rolling(24).mean()


# Remove rows with missing values
df = df.dropna().reset_index(drop=True)

print("New features created successfully!")
print("Dataset shape:", df.shape)
```

```python
target = "OT_future_24"

X = df.drop(columns=["date", target])
y = df[target]

split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_improved = RandomForestRegressor(
    n_estimators=300,
    max_depth=15,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

rf_improved.fit(X_train, y_train)

y_pred_improved = rf_improved.predict(X_test)

print("Improved Random Forest training completed!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

improved_mae = mean_absolute_error(y_test, y_pred_improved)
improved_rmse = np.sqrt(mean_squared_error(y_test, y_pred_improved))
improved_r2 = r2_score(y_test, y_pred_improved)

print("Improved MAE:", improved_mae)
print("Improved RMSE:", improved_rmse)
print("Improved R2:", improved_r2)
```

```python
comparison_features = pd.DataFrame({
    "Model": [
        "Baseline",
        "Random Forest",
        "Gradient Boosting",
        "RF With Thermal Features"
    ],
    "MAE": [
        baseline_mae,
        mae,
        gb_mae,
        improved_mae
    ],
    "RMSE": [
        baseline_rmse,
        rmse,
        gb_rmse,
        improved_rmse
    ],
    "R2": [
        baseline_r2,
        r2,
        gb_r2,
        improved_r2
    ]
})

comparison_features.sort_values("MAE")
```

# Stop random experiments, start systematic tuning

```python
# Use the original feature
original_features = [
    "OT",
    "OT_lag_1",
    "OT_lag_6",
    "OT_lag_12",
    "OT_lag_24",
    "OT_lag_48",
    "HUFL",
    "HULL",
    "MUFL",
    "MULL",
    "LUFL",
    "LULL",
    "HUFL_lag_24",
    "HULL_lag_24",
    "MUFL_lag_24",
    "MULL_lag_24",
    "LUFL_lag_24",
    "LULL_lag_24"
]

X = df[original_features]
y = df["OT_future_24"]

split_index = int(len(df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]
```

```python
# Time-series cross-validation
from sklearn.model_selection import TimeSeriesSplit

tscv = TimeSeriesSplit(n_splits=5)

print("Time-series cross-validation ready!")
```

```python
# Tune Random Forest
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV

rf = RandomForestRegressor(random_state=42, n_jobs=-1)

param_grid = {
    "n_estimators": [100, 200, 300, 500],
    "max_depth": [5, 10, 15, 20, None],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4, 8],
    "max_features": [0.5, 0.7, 1.0]
}

search = RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_grid,
    n_iter=20,
    scoring="neg_mean_absolute_error",
    cv=tscv,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

search.fit(X_train, y_train)

print("Best parameters:", search.best_params_)
print("Best CV MAE:", -search.best_score_)
```

```python
# Evaluate tuned model on test set
best_rf = search.best_estimator_

y_pred_tuned = best_rf.predict(X_test)

tuned_mae = mean_absolute_error(y_test, y_pred_tuned)
tuned_rmse = np.sqrt(mean_squared_error(y_test, y_pred_tuned))
tuned_r2 = r2_score(y_test, y_pred_tuned)

print("Tuned MAE:", tuned_mae)
print("Tuned RMSE:", tuned_rmse)
print("Tuned R2:", tuned_r2)
```

```python
# Fair comparison — run this now
y_pred_baseline_current = X_test["OT"].values

baseline_mae_current = mean_absolute_error(y_test, y_pred_baseline_current)
baseline_rmse_current = np.sqrt(mean_squared_error(y_test, y_pred_baseline_current))
baseline_r2_current = r2_score(y_test, y_pred_baseline_current)

print("Current Baseline MAE:", baseline_mae_current)
print("Current Baseline RMSE:", baseline_rmse_current)
print("Current Baseline R2:", baseline_r2_current)
```

```python
comparison_tuned = pd.DataFrame({
    "Model": ["Baseline", "Tuned Random Forest"],
    "MAE": [baseline_mae_current, tuned_mae],
    "RMSE": [baseline_rmse_current, tuned_rmse],
    "R2": [baseline_r2_current, tuned_r2]
})

comparison_tuned.sort_values("MAE")
```
